## Vector Spaces

When we first introduced vectors back in [Chapter 2.1](01-vectors.ipynb), I mentioned that they support two key operations: addition and scalar multiplication. These two operations combine through **linear combinations**, as we've seen repeatedly throughout Chapter 2. It turns out that we can take linear combinations of mathematical objects other than vectors.

:::{note} Definition: Vector Space

A **vector space** is a set of objects $V$ where:
1. We can add any two elements of $V$ and the result is also an element of $V$, i.e. if $\vec u, \vec v \in V$, then $\vec u + \vec v \in V$.
2. We can multiply any element of $V$ by a scalar $c$ and the result is also an element of $V$, i.e. if $\vec u \in V$ and $c$ is a scalar, then $c \vec u \in V$.
:::

$\mathbb{R}^2$ is a vector space: if you take any two vectors in $\mathbb{R}^2$ and add them together, you will still have a vector in $\mathbb{R}^2$, and likewise for scalar multiplication. $\mathbb{R}^3$, $\mathbb{R}^4$, and more generally, $\mathbb{R}^n$ are all vector spaces, too.

What might be a vector space that isn't made up of "vectors" in the traditional sense? How about the set of **polynomials with degree $\leq 3$**? (3 was chosen arbitrarily here.)

$$u(x) = -2x^2 + 7x - 1$$

$$v(x) = 3x^3 - 4x^2 + 2x + 5$$

In the vector space of polynomials with degree $\leq 3$, $u(x)$ and $v(x)$ are vectors. Why is this a vector space? If you add any two polynomials with degree $\leq 3$ together, you will still have a polynomial with degree $\leq 3$. Likewise, if you multiply a polynomial with degree $\leq 3$ by a scalar, you will still have a polynomial with degree $\leq 3$.

Why are the set of polynomials of degree **exactly** 3 **not** a vector space?
:::

There are other examples of abstract vector spaces, too, like the set of all **matrices** with real entries, or the set of all continuous functions from $\mathbb{R}$ to $\mathbb{R}$. In the definition of a vector space, I mentioned that the objects in the space should support addition and scalar multiplication. We know how these operations work for our traditional definition of vectors, but what do these operations look like for generic mathematical objects?

:::{note} Vector Space Axioms
In order for a set of objects $V$ to be a vector space, their definitions of addition and scalar multiplication must satisfy the following 8 properties, for elements $\vec u, \vec v, \vec w \in V$ and scalars $c, d \in \mathbb{R}$.

Regarding addition:
1. $\vec u + \vec v = \vec v + \vec u$ (addition is commutative)
2. $(\vec u + \vec v) + \vec w = \vec u + (\vec v + \vec w)$ (addition is associative)
3. There exists an element $\vec 0 \in V$ such that $\vec v + \vec 0 = \vec v$ for all $\vec v \in V$ (existence of an additive identity)
4. For every $\vec v \in V$, there exists an element $-\vec v \in V$ such that $\vec v + (-\vec v) = \vec 0$ (existence of additive inverses)

Regarding scalar multiplication:

5. $1 \vec u = \vec u$ (existence of a multiplicative identity)
6. $(cd)\vec v = c(d \vec v)$ (scalar multiplication is associative)

Regarding both:

7. $c(\vec u + \vec v) = c\vec u + c\vec v$ (scalar multiplication distributes over vector addition)
8. $(c + d)\vec v = c\vec v + d\vec v$ (scalar addition distributes over scalar multiplication)
:::

Technically, we could be even more abstract and not restrict the scalars to be real numbers, and instead allow them to be complex numbers. We'll avoid that possibility for now, and in general, stick to focusing on the vector space $\mathbb{R}^n$. (Still, I wanted to provide you with some context for how to generalize these ideas.)

:::{tip} Activity 2
Suppose we _try_ to create a new vector space in which vector addition is defined in the following way:

$${\color{orange}\begin{bmatrix} u_1 \\ u_2 \end{bmatrix}} + {\color{#3d81f6} \begin{bmatrix} v_1 \\ v_2 \end{bmatrix}} = \begin{bmatrix} {\color{orange}u_1} + {\color{#3d81f6}v_2} \\ {\color{orange}u_2} + {\color{#3d81f6}v_1} \end{bmatrix}$$

## Subspaces

As I said earlier, we'll stick to considering vector spaces of the form $\mathbb{R}^n$. 

A theme that we've studied over the past few sections is that of a subspace: a subset of a vector space. The span of a set of vectors in $\mathbb{R}^n$ is a subspace of $\mathbb{R}^n$, as is a line or plane through the origin. Let me be a bit more precise.

:::{note} Definition: Subspace

A subspace $S$ of a vector space $V$ is a set of vectors in $V$ where:

1. The zero vector is in $S$
1. If $\vec u, \vec v \in S$, then $\vec u + \vec v \in S$
1. If $\vec u \in S$ and $c$ is a scalar, then $c \vec u \in S$

The last two properties together tell us that **$S$ must be "closed" under addition and scalar multiplication**.
:::

A simpler way of thinking of the last two properties is that if you take any two vectors $\vec u, \vec v \in S$, then any linear combination $c\vec u + d\vec v$ must also be in $S$. This is the practical way we'll "check" if $S$ is a subspace of $V$.

Many years ago, one of the great math teachers I had used the following example to explain what the term **closure** meant:

> If you put two elephants in a cage, and one day a third animal comes out, that third animal is still an elephant. - [J. Bruce White](https://www.familiesfirst.ca/memorials/j-bruce-white/5608620/), 1938-2025

The plane below is an example of a subspace of $\mathbb{R}^3$.

```python
import numpy as np
import plotly.graph_objects as go

v1 = (5, 2, 1)
v2 = (-2, 3, 0)

plane_extent = 20
num_points = 3

s_range = np.linspace(-plane_extent, plane_extent, num_points)
t_range = np.linspace(-plane_extent, plane_extent, num_points)
s_grid, t_grid = np.meshgrid(s_range, t_range)

plane_x = s_grid * v1[0] + t_grid * v2[0]
plane_y = s_grid * v1[1] + t_grid * v2[1]
plane_z = s_grid * v1[2] + t_grid * v2[2]

fig = go.Figure()

fig.add_trace(go.Surface(
    x=plane_x,
    y=plane_y,
    z=plane_z,
    opacity=0.8,
    colorscale=[[0, 'rgba(61,129,246,0.8)'], [1, 'rgba(61,129,246,0.8)']],
    showscale=False,
))

# Add faint lines at x=0, y=0, z=0
line_range = [-10, 10]
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=line_range, z=[0, 0],
    mode='lines',
    line=dict(color='#f0f0f0', width=2),
    showlegend=False,
    hoverinfo='skip'
))
fig.add_trace(go.Scatter3d(
    x=line_range, y=[0, 0], z=[0, 0],
    mode='lines',
    line=dict(color='#f0f0f0', width=2),
    showlegend=False,
    hoverinfo='skip'
))
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=line_range,
    mode='lines',
    line=dict(color='#f0f0f0', width=2),
    showlegend=False,
    hoverinfo='skip'
))

fig.update_layout(
    scene=dict(
        xaxis=dict(
            range=[-5, 5],
            title='x₁',
            backgroundcolor='white',
            gridcolor='#f0f0f0',
            zerolinecolor='#f0f0f0',
            showbackground=True,
            # tickfont=dict(family='Palatino'),
            # titlefont=dict(family='Palatino')
        ),
        yaxis=dict(
            range=[-5, 5],
            title='x₂',
            backgroundcolor='white',
            gridcolor='#f0f0f0',
            zerolinecolor='#f0f0f0',
            showbackground=True,
            # tickfont=dict(family='Palatino'),
            # titlefont=dict(family='Palatino')
        ),
        zaxis=dict(
            range=[-5, 5],
            title='x₃',
            backgroundcolor='white',
            gridcolor='#f0f0f0',
            zerolinecolor='#f0f0f0',
            showbackground=True,
            # tickfont=dict(family='Palatino'),
            # titlefont=dict(family='Palatino')
        ),
        aspectmode='cube',
    ),
    width=600,
    height=600,
    scene_camera=dict(
        eye=dict(x=-1.2, y=2, z=1.2)
    ),
    font=dict(family='Palatino'),
    paper_bgcolor='white',
    plot_bgcolor='white'
)

fig.show()
```

Pick any two vectors that lie in the plane (when you draw them starting at the origin). Take any linear combination of those two vectors, and you'll still be in the plane. Why? We've seen now that planes through the origin are objects that are spanned by two vectors, and by definition this plane is the set of all linear combinations of those two vectors. So, any linear combination of vectors in the plane will still be in the plane, meaning the plane is a subspace.

:::{attention} Spans are Subspaces!
If $\vec v_1, \vec v_2, \ldots, \vec v_d$ are vectors in $\mathbb{R}^n$, then

$$\text{span}(\{\vec v_1, \vec v_2, \ldots, \vec v_d\}) = \left\{c_1\vec v_1 + c_2\vec v_2 + \cdots + c_d\vec v_d \;\middle\vert\; c_1, c_2, \ldots, c_d \in \mathbb{R}\right\}$$

is a subspace of $\mathbb{R}^n$.
:::

## Basis and Dimension

:::{note} Definition: Basis
A **basis** for a subspace $S$ is a set of vectors that:
1. span $S$
2. are linearly independent
:::

Think of a basis as the minimum set of vectors you need to describe every vector in a subspace:
- If you remove a vector from a basis, you'll no longer be able to reach every vector in the subspace.
- If you add an extra vector to a basis, the new set will no longer be linearly independent.

This is not a new idea: in [Chapter 2.4](04-span-linear-independence.ipynb), we looked at how to find linearly independent subsets with the same span as a given set of vectors, and all we were doing there was finding a basis for the span of the given set of vectors.
